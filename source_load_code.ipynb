```python
source_dir = '/Volumes/incremental_load_catalog/default/daily-data-volume/source/'
target_dir = '/Volumes/incremental_load_catalog/default/daily-data-volume/archive/'
stage_table = 'incremental_load_catalog.default.staging_table'
```

```python
df = spark.read.format('csv').option('header', 'true').option('inferSchema', 'true').load(source_dir)

display(df)
```

```python
df.write.format('delta').mode('overwrite').saveAsTable(stage_table)
```

```python
files = dbutils.fs.ls(source_dir)

for file in files:
    src_path = file.path

    target_path = target_dir + src_path.split('/')[-1]
    dbutils.fs.mv(src_path, target_path)
```
